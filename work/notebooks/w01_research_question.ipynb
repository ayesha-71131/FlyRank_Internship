## 1. My lane: Lane 2 — Refresh / Content Opportunity Scoring

Most content decays over time, and SEO teams have limited capacity to review every page. I want to build a scoring system that ranks which pages need attention first, using observable signals like position, engagement, and historical performance, so a strategist can focus their limited time on pages with the highest opportunity or risk.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the dataset from sample_data
df = pd.read_csv('/content/sample_data/content_refresh_anonymized.csv')

# Number 1: Most content is declining
declining_pct = (df['trend_direction'] == 'down').mean() * 100
print(f"1. Pages trending down: {declining_pct:.1f}% — most content needs attention")

# Number 2: Past performance predicts future performance (traffic is sticky)
corr_past_future = df['impressions_prev_30d'].corr(df['impressions_last_30d'])
print(f"2. Past vs future impressions correlation: {corr_past_future:.3f} — traffic patterns persist")

# Number 3: Search volume doesn't predict traffic (simple rules fail)
corr_volume_impressions = df['search_volume'].corr(df['impressions_90d'])
print(f"3. Search volume vs impressions correlation: {corr_volume_impressions:.3f} — simple rules won't work")

print("These numbers show: content decays, historical patterns matter,")
print("and search volume alone is useless — we need a better scoring system.")

```

## 2. The question: decision, action, cost of a wrong call

**Decision:** Which pages should an SEO strategist review first for refresh, expansion, or pruning.

**Who acts:** The SEO strategist or content manager uses the ranked queue to decide which pages to work on each week.

**Cost of wrong call:**

**False positive** (recommending a page that doesn't need help): Wastes limited strategist hours on pages that would have performed fine anyway.

**False negative** (missing a page that needs help): Content continues to decay, losing traffic and revenue that could have been saved with timely intervention

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# Pages with significant traffic that are declining — high cost if missed
high_volume_declining = df[(df['impressions_90d'] >= 1000) & (df['trend_direction'] == 'down')]
print(f"1. Pages with 1000+ impressions that are declining: {len(high_volume_declining):,}")
print(f"   → If we miss these, we lose significant traffic")

# Pages that look good but might be wasting time if recommended
low_volume_declining = df[(df['impressions_90d'] < 100) & (df['trend_direction'] == 'down')]
print(f"\n2. Pages with <100 impressions that are declining: {len(low_volume_declining):,}")
print(f"   → If we recommend these, strategists waste time on low-impact pages")

# Average impressions at risk per declining page
avg_impressions_at_risk = df[df['trend_direction'] == 'down']['impressions_90d'].mean()
print(f"\n3. Average impressions per declining page: {avg_impressions_at_risk:.0f}")
print(f"   → Each wrong miss costs this much potential traffic")

print("A good scoring system helps prioritize high-impact pages")
print("while avoiding wasted effort on low-impact ones.")
```

## 4. Careful words: what I can and can't claim

**What I CAN claim:**

trends) are associated with whether a page is likely to decline or maintain traffic

That my recommendations are decision-support — not guarantees, but evidence-based suggestions

**What I CANNOT claim:**

That I've proven what causes a page to decline or recover (correlation ≠ causation)

That I know how Google's algorithm works or what factors it uses

That refreshing a recommended page will definitely improve its traffic

That my model "predicts the future" with certainty — only that it identifies patterns worth investigating
